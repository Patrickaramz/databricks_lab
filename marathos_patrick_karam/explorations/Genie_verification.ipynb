```python
# Verifierar könsfördelning i dim_athlete
from pyspark.sql.functions import col, count, round

df_verification = (
    spark.read.table("marathos.gold.dim_athlete")
    .groupBy("Athlete gender")
    .agg(count("*").alias("number_of_athletes"))
    .orderBy(col("number_of_athletes").desc())
)

display(df_verification)
```

```python
# Topp 5 länder med mest löpare
df_top_countries = spark.sql("""
    SELECT `Athlete country`, COUNT(*) as total_athletes
    FROM marathos.gold.dim_athlete
    WHERE `Athlete country` IS NOT NULL
    GROUP BY `Athlete country`
    ORDER BY total_athletes DESC
    LIMIT 5
""")

display(df_top_countries)
```

```python
# Antal event per år
df_events_per_year = spark.sql("""
    SELECT 
        `Year of event`, 
        COUNT(*) as total_events
    FROM marathos.gold.dim_event
    GROUP BY `Year of event`
    ORDER BY `Year of event` DESC
    LIMIT 10
""")

display(df_events_per_year)
```
