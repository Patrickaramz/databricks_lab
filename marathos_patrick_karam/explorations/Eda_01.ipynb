```python
# Läser in rådata
file_path = "/Volumes/marathos/default/raw/TWO_CENTURIES_OF_UM_RACES.csv"

# Läser in csv med inferred schema och header
df = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load(file_path)
    
)

# Visar de första raderna
display(df.limit(10))
```

```python
# Räknar och skriver ut antaler rader och kolumner i datasetet
num_rows = df.count()
num_cols = len(df.columns)

print(f"Antal rader i datasetet: {num_rows}")
print(f"Antal kolumner i datasetet: {num_cols}")
```

```python
# Skriver ut datatyperna för att inspektera schemat
df.printSchema()
```

```python
# Skriver ut statistik (count, mean, stddev, min, max)
display(df.summary())
```

```python
from pyspark.sql.functions import col, count, when

# Räkna ut antalet nulls i varje kolumn
null_counts = df.select(
    [count(when(col(c).isNull(), c)).alias(c) for c in df.columns]
)

display(null_counts)
```

```python
# Kontrollerar antal unika tävlingar/event
unique_events = df.select("Event name").distinct().count()
print(f"Antal unika tävlingar/event: {unique_events}")
```

```python
# Ålderfördelning baserat på ålderkategorier/födelseår
if "Athlete year of birth" in df.columns:
    display(
        df.groupBy("Athlete year of birth")
        .count()
        .orderBy("Athlete year of birth")
    )
```

```python
# Bäst presterande länder baserat på löparnas nationalitet
if "Athlete country" in df.columns:
    display(
        df.groupBy("Athlete country")
        .count()
        .orderBy(col("count").desc())
    )
```
