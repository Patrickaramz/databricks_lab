```sql
-- Skapar katalogen marathos
CREATE CATALOG IF NOT EXISTS marathos;
USE CATALOG marathos;

-- Skapar scheman (bronze, silver, gold, default)
CREATE SCHEMA IF NOT EXISTS bronze;
CREATE SCHEMA IF NOT EXISTS silver;
CREATE SCHEMA IF NOT EXISTS gold;
CREATE SCHEMA IF NOT EXISTS default;

-- Skapra en Volume för rådata under default-schemat
CREATE VOLUME IF NOT EXISTS marathos.default.raw;
```
