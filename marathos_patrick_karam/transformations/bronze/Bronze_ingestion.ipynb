```python
from pyspark.sql.functions import current_timestamp

#  Sökväg till rådata i Volume
file_path = "/Volumes/marathos/default/raw/TWO_CENTURIES_OF_UM_RACES.csv"

#  Läser in rådatan
df_raw = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load(file_path)
)

#  Lägg till ingestion_time för spårbarhet
df_bronze = df_raw.withColumn("ingestion_time", current_timestamp())

#  Sparar som Delta-tabell i Unity Catalog med Column Mapping aktiverat
(
    df_bronze.write.format("delta")
    .option("delta.columnMapping.mode", "name")
    .mode("overwrite")
    .saveAsTable("marathos.bronze.raw_marathon")
)

print("Bronze layer har sparats i marathos.bronze.raw_marathon!")
```
